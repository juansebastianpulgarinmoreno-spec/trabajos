## 6) Código funcional **[15 pts]**
Incluye tu implementación completa.

No pegues código incompleto; debe ser ejecutable.


```python
def calcular_suscriptores():
    while 1:
        entrada = input("Introduce el numero del mes (n): ").strip()
        
        if entrada.isdigit():
            mes = int(entrada)
            break
            
        print("Error: Ingresa un numero entero valido y mayor o igual a 0.")

    memoria = [0, 1]
    while len(memoria) <= mes:
        memoria.append(memoria[-1] + memoria[-2])

    resultado = memoria[mes]
    print(f"En el mes {mes}, el numero de nuevos suscriptores sera: {resultado}")

if __name__ == "__main__":
    calcular_suscriptores()



```
